Есть два пропуска в стоблце age

```python
# Проверка на полностью дублирующиеся строки
df.duplicated().sum()
```

```python
# Отрицательные значения в числовых столбцах
numeric_cols = ['age', 'purchases', 'revenue']
for col in numeric_cols:
    negative_count = (df[col] < 0).sum()
    if negative_count > 0:
        print(f"Отрицательные значения в '{col}': {negative_count} шт.")

# Нереалистичный возраст
age_over_100 = (df['age'] > 100).sum()
if age_over_100 > 0:
    print(f"Возраст более 100 лет: {age_over_100} шт.")

# Нулевой доход при ненулевых покупках
zero_revenue_nonzero_purchases = df[(df['revenue'] == 0) & (df['purchases'] > 0)]
if not zero_revenue_nonzero_purchases.empty:
    print(f"Нулевой доход при ненулевых покупках: {len(zero_revenue_nonzero_purchases)} случая(ев)")

# Gроверим, нет ли нулевых или отрицательных покупок при положительном доходе
negative_purchases = (df['purchases'] < 0).sum()
if negative_purchases > 0:
    print(f"Отрицательное количество покупок: {negative_purchases} шт.")
```

В данных есть пропуски в столбце age — 2 значения из 41 (4.9%). 

Обнаружен один подозрительный возраст — более 100 лет, что маловероятно и может быть ошибкой.

Есть одна запись, где количество покупок отрицательное — это явно является ошибкой. 

Найден один пользователь, который совершил покупки, но его доход равен нулю.

Одна строка является полным дубликатом.

```python
# Частоты по странам
print("Распределение по странам:")
print(df['country'].value_counts())

# Создание возрастных групп (с заполнением пропусков и обработкой выбросов)
df_clean = df.copy()

# Заполняем пропуски в age медианой
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())

# Устанавливаем разумные границы возраста (исключаем > 100 как ошибку)
df_clean = df_clean[df_clean['age'] <= 100]

# Создаём возрастные группы
bins = [0, 25, 35, 50, 65]
labels = ['18-25', '26-35', '36-50', '51-65']
df_clean['age_group'] = pd.cut(df_clean['age'], bins=bins, labels=labels)

# Группировка: средние и суммарные показатели по странам
grouped = df_clean.groupby('country')[['purchases', 'revenue']].agg(['mean', 'sum'])
print("\nСредние и суммарные значения по странам:")
print(grouped)

# Агрегация по возрастным группам
print("\nСредний доход по возрастным группам:")
print(df_clean.groupby('age_group')['revenue'].mean())
```

Распределение пользователей по странам неравномерное: лидируют Россия: 13 пользователей и Франция: 12, далее следуют США: 8, Германия: 6 и Китай: 2.

По средним показателям видно, что пользователи из Германии имеют самый высокий средний доход на пользователя — около 1445, хотя их общее количество покупок ниже, чем у других. В то же время, в России наибольшая общая выручка: 10 271 за счёт большего числа пользователей, но средний чек ниже. Пользователи из возрастной группы 26–35 лет приносят наибольший средний доход: 949, немного опережая группу 36–50 лет. Самый низкий доход — у самой молодой: 18–25 и самой старшей: 51–65 групп. Несмотря на активность: количество покупок, молодёжь тратит меньше.

```python
# Гистограмма для количественного признака — age
plt.figure()
plt.hist(df_clean['age'], color='skyblue', edgecolor='black')
plt.title('Распределение возраста пользователей')
plt.xlabel('Возраст')
plt.ylabel('Число пользователей')
plt.tight_layout()
plt.savefig("figures/age_histogram.png")
plt.show()
```

Основная аудитория — молодые и зрелые взрослые (20–55 лет), с максимальной концентрацией в 30–35 лет.

```python
# Боксплот по доходу (revenue) с разбивкой по странам
plt.figure(figsize=(10, 6))
df_clean.boxplot(column='revenue', by='country', ax=plt.gca())
plt.title('Распределение дохода по странам')
plt.suptitle('')  # Убираем автоматический заголовок от boxplot
plt.ylabel('Доход')
plt.xlabel('Страна')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig("figures/revenue_boxplot.png")
plt.show()
```

Пользователи из Германии приносят наибольший средний доход, а также имеют самых «дорогих» покупателей.

```python
# Scatter plot: зависимость revenue от purchases
plt.figure(figsize=(8, 6))
for country in df_clean['country'].unique():
    subset = df_clean[df_clean['country'] == country]
    plt.scatter(subset['purchases'], subset['revenue'], label=country, alpha=0.7, s=50)

plt.title('Зависимость дохода от количества покупок')
plt.xlabel('Количество покупок')
plt.ylabel('Доход')
plt.legend(title='Страна')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("figures/revenue_vs_purchases_scatter.png")
plt.show()
```

Хотя в среднем доход растёт с увеличением числа покупок, не все покупки равны по стоимости.
